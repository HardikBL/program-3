```python
import matplotlib.pyplot as plt
import csv

x = []
y = []

with open("stock_market.csv", "r") as csvfile:
    lines = csv.reader(csvfile, delimiter=",")

    for row in lines:
        x.append(row[0])
        y.append(float(row[1]))

print(x)
print(y)

plt.plot(x, y, color="darkblue", linestyle="-",
         marker="*", ms=15, mec="pink",
         label="Closing Price")

plt.xticks(rotation=45)
plt.xlabel("Date", fontsize=14, color="brown")
plt.ylabel("Closing Price", fontsize=14, color="brown")
plt.title("Stock Market Closing Price Trend",
          fontsize=20, color="g")

plt.grid()
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

rng = np.random.default_rng(42)

ec = rng.integers(800, 3000, 50)

df = pd.DataFrame({
    "Engine Capacity (cc)": ec,
    "Fuel Efficiency (km/l)": 28 - ec / 150 + rng.normal(0, 1.5, 50)
})

df.loc[3, "Fuel Efficiency (km/l)"] = np.nan

df = df.dropna()

sns.set_style("whitegrid")

sns.scatterplot(data=df,
                x="Engine Capacity (cc)",
                y="Fuel Efficiency (km/l)",
                s=70, color="green", edgecolor="black")

plt.title("Engine Capacity vs Fuel Efficiency")
plt.xlabel("Engine Capacity (cc)")
plt.ylabel("Fuel Efficiency (km/l)")

plt.tight_layout()
plt.savefig("scatter_plot.png", dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

months = pd.date_range("01-01-2023", periods=36, freq="MS")

t = np.arange(36)

admitted = (900 + 2 * t + 60 * np.sin(2 * np.pi * t / 12)).round()

emergency = (120 + 25 * np.cos(2 * np.pi * t / 12)).round()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(months, admitted, marker="^", color="red")
axes[0].set_title("Monthly Patient Admissions")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Patients Admitted")

axes[1].bar(months, emergency, width=20, color="green")
axes[1].set_title("Emergency Cases by Month")
axes[1].set_xlabel("Month")
axes[1].set_ylabel("Emergency Cases")

plt.tight_layout()
plt.show()
```
