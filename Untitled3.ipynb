```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

months = pd.date_range("01-01-2023", periods=36, freq="MS")

t = np.arange(36)

admitted = (900 + 2 * t + 60 * np.sin(2 * np.pi * t / 12)).round()

emergency = (120 + 25 * np.cos(2 * np.pi * t / 12)).round()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(months, admitted, marker="^", color="red")
axes[0].set_title("Monthly Patient Admissions")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Patients Admitted")

axes[1].bar(months, emergency, width=20, color="green")
axes[1].set_title("Emergency Cases by Month")
axes[1].set_xlabel("Month")
axes[1].set_ylabel("Emergency Cases")

plt.tight_layout()
plt.show()
```
